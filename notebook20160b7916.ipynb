```python
import os

src = "/kaggle/working/final_submission.zip"
chunk_size = 50 * 1024 * 1024  # 50 MB

with open(src, "rb") as f:
    i = 1
    while True:
        data = f.read(chunk_size)
        if not data:
            break

        chunk = f"/kaggle/working/final_submission.part{i:02d}"
        with open(chunk, "wb") as out:
            out.write(data)

        print(f"Created {chunk} - {os.path.getsize(chunk)/(1024**2):.1f} MB")
        i += 1
```

```python
import os
import shutil

# Create a dedicated folder containing only the ZIP parts
parts_dir = "/kaggle/working/download_parts"

if os.path.exists(parts_dir):
    shutil.rmtree(parts_dir)

os.makedirs(parts_dir)

for i in range(1, 11):
    src = f"/kaggle/working/final_submission.part{i:02d}"
    dst = f"{parts_dir}/final_submission.part{i:02d}"
    shutil.copy2(src, dst)

print("Parts prepared:")
for f in sorted(os.listdir(parts_dir)):
    print(f, round(os.path.getsize(os.path.join(parts_dir, f))/(1024**2), 2), "MB")
```
